## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

## Data Wrangling

### Gathering Data

Dataset Air Quality terdiri dari 12 file CSV yang merepresentasikan data pengamatan dari masing-masing stasiun pemantauan kualitas udara di Beijing. Setiap file memiliki struktur kolom yang sama, sehingga seluruh data dapat digabungkan menjadi satu dataframe untuk memudahkan proses pengolahan dan analisis.

#### Load Dataset Air Quality

```python
zip_path = "/content/PRSA_Data_20130301-20170228.zip"
extract_path = "/content/air_quality_dataset"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)
```

```python
air_files = glob.glob(
    "/content/air_quality_dataset/**/*.csv",
    recursive=True
)

air_data = {}

for file in air_files:
    station_name = os.path.basename(file).replace(
        "PRSA_Data_", ""
    ).replace("_20130301-20170228.csv", "")

    air_data[station_name] = pd.read_csv(file)

air_df = pd.concat(
    air_data.values(),
    ignore_index=True
)

print("Jumlah file:", len(air_files))
print("Jumlah stasiun:", len(air_data))
print("Ukuran dataset:", air_df.shape)

display(air_df.head())
```

**Insight:**
- Dataset Air Quality terdiri dari 12 file CSV yang masing-masing merepresentasikan data pengamatan dari satu stasiun pemantauan kualitas udara di Beijing.

- Seluruh file berhasil digabungkan menjadi satu dataframe dengan total 420.768 baris dan 18 kolom. Dataset ini mencakup data konsentrasi polutan dan kondisi meteorologi selama periode Maret 2013 hingga Februari 2017.

### Assessing Data
Pada tahap ini, dilakukan pemeriksaan terhadap dataset Air Quality untuk mengetahui kualitas dan kondisi data sebelum digunakan dalam analisis. Pemeriksaan mencakup tipe data pada setiap kolom, keberadaan nilai yang hilang (missing values), serta data duplikat. Hasil pemeriksaan tersebut digunakan untuk mengidentifikasi permasalahan pada dataset yang perlu ditangani pada tahap data cleaning.

#### Identifying Data Quality Problem

Pemeriksaan kualitas data dilakukan untuk mengetahui kondisi dataset, termasuk tipe data, nilai yang hilang (missing values), dan data duplikat. Pemeriksaan ini bertujuan untuk mengidentifikasi masalah yang perlu ditangani sebelum data digunakan dalam analisis.

```python
# Memeriksa informasi umum dataset
air_df.info()

# Memeriksa jumlah missing values pada setiap kolom
missing_values = air_df.isnull().sum()

# Menghitung persentase missing values
missing_percentage = (missing_values / len(air_df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage (%)": missing_percentage
})

display(missing_summary[missing_summary["Missing Values"] > 0])

# Memeriksa jumlah data duplikat
print("Jumlah data duplikat:", air_df.duplicated().sum())
```

```python
# Memeriksa nilai ekstrem menggunakan metode IQR
for column in ["PM2.5", "PM10"]:
    Q1 = air_df[column].quantile(0.25)
    Q3 = air_df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = air_df[
        (air_df[column] < lower_bound) |
        (air_df[column] > upper_bound)
    ]

    print(f"\n{column}")
    print("Batas bawah:", lower_bound)
    print("Batas atas:", upper_bound)
    print("Jumlah outlier:", len(outliers))
```

```python
# Visualisasi outlier pada PM2.5 dan PM10
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(y=air_df["PM2.5"], ax=axes[0])
axes[0].set_title("Boxplot PM2.5")
axes[0].set_ylabel("Konsentrasi PM2.5")

sns.boxplot(y=air_df["PM10"], ax=axes[1])
axes[1].set_title("Boxplot PM10")
axes[1].set_ylabel("Konsentrasi PM10")

plt.tight_layout()
plt.show()
```

**Steps to Take:**
- Menangani nilai yang hilang (missing values) pada kolom pengukuran polutan dan kondisi meteorologi sesuai kebutuhan analisis.
- Memastikan format data waktu dapat digunakan untuk analisis berdasarkan tanggal dan jam.
- Mempertahankan data yang tersedia karena tidak ditemukan baris duplikat pada dataset.
- Memeriksa nilai ekstrem (outlier) pada kolom PM2.5 dan PM10 menggunakan metode IQR. Nilai yang terdeteksi sebagai outlier akan dipertahankan karena belum tentu merupakan kesalahan pencatatan.

**Insight:**
- Dataset memiliki 420.768 baris dan 18 kolom dengan tipe data yang secara umum sudah sesuai.
- Ditemukan nilai yang hilang pada 12 kolom, dengan persentase tertinggi pada kolom CO sebesar 4,92%. Sementara itu, tidak ditemukan data duplikat pada dataset.
- Pemeriksaan IQR menemukan 19.142 outlier pada PM2.5 dan 14.658 outlier pada PM10.
- Hasil boxplot menunjukkan adanya nilai ekstrem pada kedua kolom tersebut, tetapi nilai-nilai tersebut tetap dipertahankan karena belum tentu merupakan kesalahan pencatatan.

### Cleaning Data
Pada tahap ini, dilakukan pembersihan dan penyesuaian data berdasarkan permasalahan yang ditemukan saat proses assessing. Penanganan difokuskan pada nilai yang hilang serta penambahan kolom waktu (datetime) untuk mendukung analisis berdasarkan periode pengamatan.

#### Fixing Data Format and Handling Missing Values

Pada tahap ini, dilakukan penyesuaian format data waktu dengan menggabungkan kolom year, month, day, dan hour menjadi satu kolom datetime. Missing values tetap dipertahankan tanpa imputasi maupun penghapusan karena penanganannya disesuaikan dengan kebutuhan masing-masing analisis. Selain itu, nilai ekstrem (outlier) pada kolom PM2.5 dan PM10 juga dipertahankan karena belum dapat dipastikan sebagai kesalahan pencatatan dan berpotensi merepresentasikan kondisi kualitas udara yang sebenarnya.

```python
# Menggabungkan kolom tahun, bulan, tanggal, dan jam
# menjadi satu kolom datetime
air_df["datetime"] = pd.to_datetime(
    air_df[["year", "month", "day", "hour"]]
)

# Memeriksa hasil pembuatan kolom datetime
display(air_df[["year", "month", "day", "hour", "datetime"]].head())
```

```python
# Memeriksa kembali jumlah missing values
print("Jumlah missing values:")
display(air_df.isnull().sum()[air_df.isnull().sum() > 0])

# Memeriksa tipe data kolom datetime
print("\nTipe data datetime:", air_df["datetime"].dtype)

# Memeriksa jumlah data duplikat
print("Jumlah data duplikat:", air_df.duplicated().sum())
```

**Insight:**
- Kolom datetime berhasil dibuat dengan menggabungkan kolom year, month, day, dan hour. Kolom ini memiliki tipe data datetime64[ns] sehingga dapat digunakan untuk analisis berdasarkan waktu.
- Missing values tetap dipertahankan tanpa imputasi maupun penghapusan karena penanganannya disesuaikan dengan kebutuhan masing-masing analisis.
- Nilai ekstrem (outlier) pada PM2.5 dan PM10 tetap dipertahankan karena belum dapat dipastikan sebagai data yang tidak valid dan berpotensi merepresentasikan kondisi kualitas udara sebenarnya.

**Insight:**

```python
# Mengelompokkan kecepatan angin berdasarkan kuartil
air_df["wind_category"] = pd.qcut(
    air_df["WSPM"],
    q=4,
    labels=["Rendah", "Sedang", "Tinggi", "Sangat Tinggi"]
)

# Menghitung rata-rata PM2.5 berdasarkan kategori kecepatan angin
wind_pm25 = air_df.groupby(
    "wind_category", observed=True
)["PM2.5"].mean()

display(wind_pm25)
```

**Insight:**
- Rata-rata PM2.5 tertinggi terdapat pada kategori kecepatan angin rendah, yaitu 102,30.
- Kategori kecepatan angin sangat tinggi memiliki rata-rata PM2.5 terendah, yaitu 45,11.
- Pola ini menunjukkan adanya hubungan berlawanan antara kecepatan angin dan konsentrasi PM2.5 pada dataset. Namun, hasil ini belum membuktikan hubungan sebab-akibat.

```python
# Membuat kategori kondisi curah hujan
air_df["rain_category"] = pd.Series(
    pd.NA, index=air_df.index, dtype="object"
)

air_df.loc[air_df["RAIN"] == 0, "rain_category"] = "Tidak Hujan"
air_df.loc[air_df["RAIN"] > 0, "rain_category"] = "Hujan"

# Menghitung rata-rata PM2.5 berdasarkan kondisi curah hujan
rain_pm25 = air_df.groupby(
    "rain_category"
)["PM2.5"].mean()

display(rain_pm25)
```

**Insight:**
- Rata-rata PM2.5 saat tidak hujan sebesar 80,50, lebih tinggi dibandingkan saat hujan sebesar 62,39.
- Hasil eksplorasi menunjukkan bahwa konsentrasi PM2.5 cenderung lebih rendah pada pengamatan saat hujan. Namun, pola tersebut tidak dapat langsung diartikan sebagai hubungan sebab-akibat.

Visualisasi ini menampilkan pola rata-rata konsentrasi PM2.5 berdasarkan jam dan bulan selama periode pengamatan. Kedua pola disajikan dalam satu gambar dengan dua grafik terpisah untuk melihat perubahan konsentrasi PM2.5 sepanjang hari dan variasinya antarbulan.

## Analisis Lanjutan (Opsional)

### Correlation Analysis
Analisis korelasi dilakukan untuk mengetahui hubungan antara konsentrasi PM2.5 dan PM10 dengan beberapa variabel meteorologi, yaitu kecepatan angin (WSPM) dan curah hujan (RAIN). Analisis ini menggunakan koefisien korelasi Pearson untuk melihat arah dan kekuatan hubungan linear antarvariabel.

```python
# Memilih kolom yang akan dianalisis
correlation_data = air_df[
    ["PM2.5", "PM10", "WSPM", "RAIN"]
]

# Menghitung korelasi antarvariabel
correlation_matrix = correlation_data.corr()

# Membuat visualisasi heatmap korelasi
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Korelasi PM2.5, PM10, dan Faktor Meteorologi")
plt.tight_layout()
plt.show()
```

**Insight Analisis Lanjutan:**
- Hasil analisis korelasi Pearson menunjukkan hubungan positif yang sangat kuat antara PM2.5 dan PM10 dengan koefisien korelasi sebesar 0,88. Hal ini menunjukkan bahwa peningkatan konsentrasi PM2.5 cenderung diikuti oleh peningkatan konsentrasi PM10.
- Kecepatan angin (WSPM) memiliki korelasi negatif dengan PM2.5 sebesar -0,27 dan PM10 sebesar -0,18. Hubungan ini menunjukkan bahwa kecepatan angin yang lebih tinggi cenderung berkaitan dengan konsentrasi polutan yang lebih rendah.
- Curah hujan (RAIN) memiliki korelasi yang sangat lemah dengan PM2.5 sebesar -0,01 dan PM10 sebesar -0,03. Hal ini menunjukkan bahwa hubungan linear antara curah hujan dan konsentrasi polutan hampir tidak terlihat.

**Catatan interpretasi:** Hasil korelasi curah hujan yang mendekati nol tidak bertentangan dengan hasil visualisasi sebelumnya. Pada analisis sebelumnya, curah hujan dikelompokkan menjadi kondisi Hujan dan Tidak Hujan, sedangkan korelasi Pearson mengukur hubungan linear berdasarkan nilai curah hujan. Keduanya menggunakan pendekatan yang berbeda.

### Manual Grouping Stasiun Berdasarkan Konsentrasi PM2.5
Analisis lanjutan dilakukan menggunakan teknik manual grouping untuk mengelompokkan stasiun pemantauan berdasarkan rata-rata konsentrasi PM2.5. Pengelompokan dibagi menjadi tiga kategori, yaitu Rendah, Sedang, dan Tinggi, berdasarkan distribusi nilai rata-rata PM2.5 pada seluruh stasiun. Analisis ini bertujuan untuk mengidentifikasi kelompok stasiun berdasarkan tingkat konsentrasi PM2.5 sehingga dapat diketahui stasiun yang memerlukan perhatian lebih dalam pemantauan kualitas udara.

```python
# Menghitung rata-rata PM2.5 berdasarkan stasiun
station_pm25 = air_df.groupby("station")["PM2.5"].mean().reset_index()

# Menentukan batas kategori berdasarkan kuantil
q1 = station_pm25["PM2.5"].quantile(1/3)
q2 = station_pm25["PM2.5"].quantile(2/3)

# Mengelompokkan stasiun ke dalam tiga kategori
station_pm25["Kategori"] = pd.cut(
    station_pm25["PM2.5"],
    bins=[-np.inf, q1, q2, np.inf],
    labels=["Rendah", "Sedang", "Tinggi"]
)

# Menampilkan hasil pengelompokan
display(station_pm25.sort_values("PM2.5"))
```

```python
# Mengurutkan data berdasarkan rata-rata PM2.5
station_pm25_sorted = station_pm25.sort_values("PM2.5")

# Membuat visualisasi hasil pengelompokan
plt.figure(figsize=(12, 6))

sns.barplot(
    data=station_pm25_sorted,
    x="station",
    y="PM2.5",
    hue="Kategori"
)

plt.title("Pengelompokan Stasiun Berdasarkan Rata-rata PM2.5")
plt.xlabel("Stasiun Pemantauan")
plt.ylabel("Rata-rata Konsentrasi PM2.5")
plt.xticks(rotation=45)
plt.legend(title="Kategori")
plt.tight_layout()
plt.show()# Mengurutkan data berdasarkan rata-rata PM2.5
station_pm25_sorted = station_pm25.sort_values("PM2.5")

# Membuat visualisasi hasil pengelompokan
plt.figure(figsize=(12, 6))

sns.barplot(
    data=station_pm25_sorted,
    x="station",
    y="PM2.5",
    hue="Kategori"
)

plt.title("Pengelompokan Stasiun Berdasarkan Rata-rata PM2.5")
plt.xlabel("Stasiun Pemantauan")
plt.ylabel("Rata-rata Konsentrasi PM2.5")
plt.xticks(rotation=45)
plt.legend(title="Kategori")
plt.tight_layout()
plt.show()
```

**Insight:**
- Hasil pengelompokan menggunakan teknik manual grouping menunjukkan bahwa 12 stasiun pemantauan terbagi secara seimbang ke dalam tiga kategori konsentrasi PM2.5, yaitu Rendah, Sedang, dan Tinggi, dengan masing-masing kategori terdiri dari empat stasiun.
- Kategori Rendah terdiri dari Dingling, Huairou, Changping, dan Shunyi dengan rata-rata PM2.5 berkisar antara 65,99 hingga 79,49. Dingling mencatatkan konsentrasi terendah sebesar 65,99.
- Kategori Sedang terdiri dari Tiantan, Aotizhongxin, Guanyuan, dan Wanliu dengan rata-rata PM2.5 berkisar antara 82,16 hingga 83,37.
- Kategori Tinggi terdiri dari Gucheng, Nongzhanguan, Wanshouxigong, dan Dongsi dengan rata-rata PM2.5 berkisar antara 83,85 hingga 86,19. Dongsi menjadi stasiun dengan konsentrasi tertinggi sebesar 86,19.
